```python
# 3-3 데이터 생성

import numpy as np
import pandas as pd

def make_spiral(sample_per_class=1000, n_classes=3, noise=0.2, turns=1.2):
    x_list, y_list = [], []
    for group in range(n_classes):
        radius = np.linspace(0., 1., sample_per_class)
        span = turns * 2 * np.pi
        theta = radius * span + (2 * np.pi * group / n_classes) \
                + np.random.randn(sample_per_class) * noise
        x = (radius * np.sin(theta))[:, None]
        y = (radius * np.cos(theta))[:, None]
        labels = np.full(sample_per_class, group, dtype=np.int8)
        x_list.append(np.hstack([x, y]))
        y_list.append(labels)
    X = np.vstack(x_list).astype(np.float32)
    Y = np.hstack(y_list).astype(np.int64)
    return X, Y

np.random.seed(42)
sample_for_classes = 300            # 난이도 상향: 1000 -> 300 (표본 축소)
number_of_groups   = 3
X_np, Y_np = make_spiral(sample_for_classes, number_of_groups, noise=0.5)  # noise 0.2 -> 0.5


file_path = r'../data/ch3_spiral_data.csv'
df = pd.DataFrame(X_np, columns=['x1', 'x2'])
df['label'] = Y_np
df_shuffled = df.sample(frac=1, random_state=42).reset_index(drop=True)
df_shuffled.to_csv(file_path, index=False)
```

```python
# 9-1 ~ 10-1 

# 일반 데이터

import random
from datetime import datetime, timedelta

src_formats = [
    '%d %B %Y',     # 01 February 2026
    '%d %b %Y',     # 01 Feb 2026
    '%B %d, %Y',    # February 01, 2026
    '%b %d, %Y',    # Feb 01, 2026
    '%m/%d/%Y',     # 02/01/2026
    '%Y/%m/%d',     # 2026/02/01
    '%d-%m-%Y',     # 01-02-2026
    '%Y-%m-%d',     # 2026-02-01
]

def choice_random_dates(n_samples=1, start_date=None, end_date=None):
    if start_date is None:
        start_date = datetime(1900, 1, 1)
        end_date = datetime(2050, 12, 31)
    days_between = (end_date - start_date).days
    datetime_list = []
    for _ in range(n_samples):
        days_after = random.randrange(days_between)
        gen_date = start_date + timedelta(days=days_after)
        datetime_list.append(gen_date)
    return datetime_list

def generate_datepairs(date_list, src_formats=src_formats):
    if len(src_formats) == 0:
        raise ValueError('입력 문자열 템플릿이 비어 있습니다.')
    src_dates, tgt_dates = [], []
    for i, date in enumerate(date_list):
        format_idx = i % len(src_formats)
        src_format = src_formats[format_idx]
        src_date = date.strftime(src_format)
        tgt_date = f'{date.year}-{date.month}-{date.day}'
        src_dates.append(src_date)
        tgt_dates.append(tgt_date)
    return src_dates, tgt_dates

date_list = choice_random_dates(4000)
src_dates, tgt_dates = generate_datepairs(date_list)

# 노이지 데이터


import string

def add_random_noise(text, max_length=40):
    noise_chars = string.ascii_letters + string.digits + '!@#$%^&*()_+-=[]{}|;:,./<>?'
    date_length = len(text)
    prefix, suffix = '', ''
    remaining = max_length - date_length
    if remaining > 0:
        prefix_length = random.randint(0, remaining)
        remaining -= prefix_length
        prefix = ''.join(random.choices(noise_chars, k=prefix_length))
    if remaining > 0:
        suffix_length = random.randint(0, remaining)
        suffix = ''.join(random.choices(noise_chars, k=suffix_length))
    return prefix + text + suffix

def generate_noisy_datepairs(date_list, src_formats=src_formats):
    
    if len(src_formats) == 0:
        raise ValueError('입력 문자열 템플릿이 비어 있습니다.')
    src_dates, tgt_dates = [], []
    for i, date in enumerate(date_list):
        format_idx = i % len(src_formats)
        src_format = src_formats[format_idx]
        src_date_text = add_random_noise(date.strftime(src_format))
        tgt_date_text = f'{date.year}-{date.month}-{date.day}'
        src_dates.append(src_date_text)
        tgt_dates.append(tgt_date_text)
    return src_dates, tgt_dates

src_noisy_dates, tgt_noisy_dates = generate_noisy_datepairs(date_list)

```

```python
# 10-2


import random
from datetime import datetime, timedelta

src_formats = [
    '%d %B %Y',  '%d %b %Y',
    '%B %d, %Y', '%b %d, %Y',
    '%m/%d/%Y',  '%Y/%m/%d',
    '%d-%m-%Y',  '%Y-%m-%d',
]

month_abbr = ['', 'Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
              'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
month_full = ['', 'January', 'February', 'March', 'April', 'May', 'June',
              'July', 'August', 'September', 'October', 'November', 'December']


def days_in_month(year: int, month: int) -> int:
    if month == 2:
        is_leap = (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)
        return 29 if is_leap else 28
    return [0, 31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31][month]


def make_date_str(year: int, month: int, day: int, fmt: str) -> str:
    fmt_map = {
        '%d %B %Y':  f'{day:02d} {month_full[month]} {year}',
        '%d %b %Y':  f'{day:02d} {month_abbr[month]} {year}',
        '%B %d, %Y': f'{month_full[month]} {day:02d}, {year}',
        '%b %d, %Y': f'{month_abbr[month]} {day:02d}, {year}',
        '%m/%d/%Y':  f'{month:02d}/{day:02d}/{year}',
        '%Y/%m/%d':  f'{year}/{month:02d}/{day:02d}',
        '%d-%m-%Y':  f'{day:02d}-{month:02d}-{year}',
        '%Y-%m-%d':  f'{year}-{month:02d}-{day:02d}',
    }
    return fmt_map[fmt]


def make_valid_date_str() -> str:
    start = datetime(1900, 1, 1)
    end = datetime(2050, 12, 31)
    days = (end - start).days
    date_obj = start + timedelta(days=random.randrange(days))
    return date_obj.strftime(random.choice(src_formats))


def make_invalid_day_str() -> str:
    year = random.randint(1900, 2050)
    month = random.randint(1, 12)
    max_day = days_in_month(year, month)
    day = random.randint(max_day + 1, max_day + 5)
    return make_date_str(year, month, day, random.choice(src_formats))


def make_invalid_format_str() -> str:
    start = datetime(1900, 1, 1)
    end = datetime(2050, 12, 31)
    days = (end - start).days
    date_obj = start + timedelta(days=random.randrange(days))
    y, m, d = date_obj.year, date_obj.month, date_obj.day
    abbr, full = month_abbr[m], month_full[m]
    invalid_formats = [
        f'{d:02d} {abbr}, {y}',   f'{d:02d} {full}, {y}',
        f'{abbr} {d:02d} {y}',    f'{full} {d:02d} {y}',
        f'{d:02d}.{m:02d}.{y}',   f'{m:02d}.{d:02d}.{y}',
        f'{y}.{m:02d}.{d:02d}',   f'{m:02d}-{d:02d}-{y}',
        f'{d:02d}/{m:02d}/{y}',   f'{y}/{d:02d}/{m:02d}',
        f'{y}-{d:02d}-{m:02d}',   f'{abbr} {y}, {d:02d}',
        f'{d:02d} {m:02d} {y}',
    ]
    return random.choice(invalid_formats)


def make_invalid_date_str() -> str:
    if random.random() < 0.5:
        return make_invalid_day_str()
    return make_invalid_format_str()




N_SAMPLES = 8000

all_strings: list[str] = []
all_labels: list[int] = []
for _ in range(N_SAMPLES // 2):
    all_strings.append(make_valid_date_str())
    all_labels.append(1)   # 정상
for _ in range(N_SAMPLES // 2):
    all_strings.append(make_invalid_date_str())
    all_labels.append(0)   # 비정상

combined = list(zip(all_strings, all_labels))
random.shuffle(combined)
all_strings, all_labels = zip(*combined)

```

```python
# 10-2 유형별 데이터 생성

import string 

def gen_type_a_valid(n: int) -> list[str]:
    return [make_valid_date_str() for _ in range(n)]


def gen_type_b_comma(n: int) -> list[str]:
    out = []
    while len(out) < n:
        start = datetime(1900, 1, 1)
        end = datetime(2050, 12, 31)
        days = (end - start).days
        date_obj = start + timedelta(days=random.randrange(days))
        y, m, d = date_obj.year, date_obj.month, date_obj.day
        abbr, full = month_abbr[m], month_full[m]
        candidates = [
            f'{d:02d} {abbr}, {y}',       # 쉼표 삽입 오류 (%d %b %Y -> %d %b, %Y)
            f'{d:02d} {full}, {y}',
            f'{abbr} {d:02d} {y}',        # 쉼표 누락 (%b %d, %Y -> %b %d %Y)
            f'{full} {d:02d} {y}',
            f'{d:02d}.{m:02d}.{y}',       # 점 구분자 (%d-%m-%Y -> %d.%m.%Y)
            f'{y}.{m:02d}.{d:02d}',
        ]
        out.append(random.choice(candidates))
    return out

# 의미 없음 --> 제거
def gen_type_c_order(n: int) -> list[str]:
    out = []
    while len(out) < n:
        start = datetime(1900, 1, 1)
        end = datetime(2050, 12, 31)
        days = (end - start).days
        date_obj = start + timedelta(days=random.randrange(days))
        y, m, d = date_obj.year, date_obj.month, date_obj.day
        abbr, full = month_abbr[m], month_full[m]
        invalid_formats = [
            f'{y} {d:02d} {full}', f'{d:02d} {y} {full}',
            f'{y} {d:02d} {abbr}', f'{d:02d} {y} {abbr}',
            f'{y} {full}, {d:02d}', f'{full} {y}, {d:02d}',
            f'{y} {abbr}, {d:02d}', f'{abbr} {y}, {d:02d}',
            f'{m:02d}/{y}/{d:02d}', f'{d:02d}-{y}-{m:02d}'
        ]


        out.append(random.choice(invalid_formats))
    return out


def gen_type_d_invalid(n: int) -> list[str]:
    out = []
    while len(out) < n:
        year = random.randint(1900, 2050)
        date = random.randint(29, 99)
        month = random.randint(1, 12)
        leap_year = False
        # 윤년이 아닌 해 선택
        if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
            leap_year = True
        # 가능한 날짜 배제
        if month == 2:
            if leap_year and date == 29:
                continue
        elif month in [4, 6, 9, 11]:
            if date < 31:
                continue
        else:
            if date < 32:
                continue
        fmt = random.choice(src_formats)
        out.append(make_date_str(year, month, date, fmt))
    return out

def make_omit_leading0_date_str(year: int, month: int, day: int, fmt: str) -> str:
    fmt_map = {
        '%d %B %Y':  f'{day} {month_full[month]} {year}',
        '%d %b %Y':  f'{day} {month_abbr[month]} {year}',
        '%B %d, %Y': f'{month_full[month]} {day}, {year}',
        '%b %d, %Y': f'{month_abbr[month]} {day}, {year}',
        '%m/%d/%Y':  f'{month}/{day}/{year}',
        '%Y/%m/%d':  f'{year}/{month}/{day}',
        '%d-%m-%Y':  f'{day}-{month}-{year}',
        '%Y-%m-%d':  f'{year}-{month}-{day}',
    }
    return fmt_map[fmt]

def gen_type_e_omit0(n):
    out = []
    while len(out) < n:
        start = datetime(1900, 1, 1)
        end = datetime(2050, 12, 31)
        days = (end - start).days
        date_obj = start + timedelta(days=random.randrange(days))
        y, m, d = date_obj.year, date_obj.month, date_obj.day        
        format_id = random.randint(0, 7)
        correct_str = make_date_str(y, m, d, src_formats[format_id])
        incorrect_str = make_omit_leading0_date_str(y, m, d, src_formats[format_id])
        if correct_str == incorrect_str:
            continue
        out.append(incorrect_str)
    return out

def gen_type_f_random(n: int) -> list[str]:
    out = []
    while len(out) < n:
        ordinary_date_str = make_valid_date_str()
        gen_length = len(ordinary_date_str)
        char_pool = string.ascii_letters + string.digits + "-/, "
        random_string = "".join(random.choices(char_pool, k=gen_length))
        out.append(random_string)
    return out


N_PER_TYPE = 200
type_specs = [
    ('A. valid format', True,  gen_type_a_valid),
    ('B. errors in seperator', False, gen_type_b_comma),
    ('C. position errors in year', False, gen_type_c_order),
    ('D. invalid date', False, gen_type_d_invalid),
    ('E. omiting leading 0 in month or date', False, gen_type_e_omit0),
    ('F. random string', False, gen_type_f_random),
]

print(f'\n{"입력 유형":<33} {"정답":<6} {"분류 정확도":<10} {"평균 신뢰 수준"}')
print('-' * 77)
for label, is_valid_truth, gen in type_specs:
    samples = gen(N_PER_TYPE)
    n_correct = 0
    sum_conf = 0.0
    for s in samples:        
        pred_label, conf = predict(model, s, vocab, device)
        pred_is_valid = (pred_label == 'O')
        if pred_is_valid == is_valid_truth:
            n_correct += 1
        sum_conf += conf
    accuracy = n_correct / N_PER_TYPE * 100
    avg_conf = sum_conf / N_PER_TYPE * 100
    truth_label = 'O' if is_valid_truth else 'X'
    print(f'{label:<40} {truth_label:<8} {accuracy:>6.1f}%            {avg_conf:>6.1f}%')
```
