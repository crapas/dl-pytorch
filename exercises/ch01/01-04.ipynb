## 연습 문제 1-4

> 다음은 세 명의 학생이 네 과목의 시험을 두 번 치른 후 성적을 정리한 표이다.
>
> | 학생 | 1회차 국어 | 영어 | 수학 | 과학 | 2회차 국어 | 영어 | 수학 | 과학 |
> |---|---|---|---|---|---|---|---|---|
> | A | 55 | 48 | 97 | 97 | 60 | 5 | 82 | 75 |
> | B | 62 | 36 | 92 | 96 | 66 | 47 | 38 | 65 |
> | C | 73 | 20 | 56 | 100 | 49 | 54 | 43 | 49 |
>
> 1. 이 표의 데이터로 학생 - 회차 - 과목 순의 텐서를 만들어 보자.
> 2. 생성한 텐서를 `permute()` 메서드를 사용해 차원이 과목 - 회차 - 학생 순서인 텐서로 변형해 보자.

**학생 - 회차 - 과목 순서의 텐서를 만드려면**
- 우선 학생별 점수를 나누고
- 학생별 점수를 회차에 따라 구분하고
- 과목 순서대로 점수를 나열한 후 묶어
- 리스트를 만든 후, 이 리스트를 사용해 텐서를 생성

```python
# 1) 학생(3) - 회차(2) - 과목(4) 순서의 3차원 텐서
scores = torch.tensor([
    [[55, 48, 97, 97], [60, 5, 82, 75]],                # 학생 A의 1회차, 2회차
    [[62, 36, 92, 96], [66, 47, 38, 65]],               # 학생 B의 1회차, 2회차
    [[73, 20, 56, 100], [49, 54, 43, 49]],              # 학생 C의 1회차, 2회차
])
```

```python
# 인덱스 값  : 0     1     2     3
# 학생의 순서: A     B     C
# 회차의 순서: 1     2
# 과목의 순서: 국어  영어  수학  과학
print(f'학생 - 회차 - 과목 텐서의 형태: {scores.shape}')
print(f'B 학생 2회차 수학 점수: {scores[1, 1, 2]}')     # 넘파이 배열 스타일의 인덱싱
print(f'C 학생 2회차 국어 점수: {scores[2][1][0]}')     # 파이썬 리스트 스타일의 인덱싱
```

```python
# 인덱싱의 결과는 파이썬 정수가 아닌 텐서!(스칼라 텐서)
b_2nd_math_score = scores[1, 1, 2]
print(type(b_2nd_math_score))
print(type(b_2nd_math_score + 10))
```

```python
# 정수가 아니므로 반드시 정수를 사용해야 하는 경우 사용할 수 없음
import random
random.seed(b_2nd_math_score)               # random.seed() 함수의 인자는 정수여야 함
```

**텐서의 차원 순서를 바꾸려면?**
- `permute()` 메서드를 사용: 각 차원이 이동할 새로운 차원을 순서대로 인자로 지정
- `permute(-1, 0, 1)`를 예로 들면
    - 첫 번째 차원을 마지막(인덱스 -1, 세 번째 차원)으로 이동
    - 두 번째 차원을 첫 번째 차원(인덱스 0)으로 이동
    - 세 번째 차원을 두 번째 차원(인덱스 1)으로 이동 


**과목 - 회차 - 학생 순서로 바꾸려면**
- 학생: 차원 인덱스 0 -> 차원 인덱스 2
- 회차: 차원 인덱스 1 -> 차원 인덱스 1(변화 없음)
- 과목: 차원 인덱스 2 -> 차원 인덱스 0
- 순서가 바뀌지 않는 차원을 포함해 모든 차원의 새로운 위치를 지정해야 함


```python
# 원래 차원 순서인 (학생, 회차, 과목)의 새로운 위치를 인자로 지정
scores_by_subject = scores.permute(2, 1, 0)                     # (과목, 회차, 학생)
# 인덱스 값  : 0     1     2     3
# 학생의 순서: A     B     C
# 회차의 순서: 1     2
# 과목의 순서: 국어  영어  수학  과학
print(f'새 텐서(과목 - 회차 - 학생 순)의 형태: {scores_by_subject.shape}')
print(f'B 학생 2회차 수학 점수: {scores_by_subject[2, 1, 1]}')  # 넘파이 배열 스타일의 인덱싱
print(f'C 학생 2회차 국어 점수: {scores_by_subject[0][1][2]}')  # 파이썬 리스트 스타일의 인덱싱
```

**`transpose()` 메서드를 사용하는 경우**
- `transpose()` 메서드는 두 차원만 맞바꿀 때 사용
- 회차 차원은 바뀌지 않으므로, 다음과 같이 과목 차원(인덱스 0)과 회차 차원(인덱스 2) 교환 가능

```python
# 인덱스 0 차원과 인덱스 -1(인덱스 2인 마지막 차원)만 맞바꾸기
scores_by_subject = scores.transpose(0, -1)                     # (과목, 회차, 학생)
# 인덱스 값  : 0     1     2     3
# 학생의 순서: A     B     C
# 회차의 순서: 1     2
# 과목의 순서: 국어  영어  수학  과학
print(f'새 텐서(과목 - 회차 - 학생 순)의 형태: {scores_by_subject.shape}')
print(f'B 학생 2회차 수학 점수: {scores_by_subject[2, 1, 1]}')  # 넘파이 배열 스타일의 인덱싱
print(f'C 학생 2회차 국어 점수: {scores_by_subject[0][1][2]}')  # 파이썬 리스트 스타일의 인덱싱
```

```python
scores.permute(1, 2)            # RuntimeError 예외 발생: 인자의 개수 == 텐서의 차원
scores.permute(1, 2, 3)         # IndexError 예외 발생: 인덱스 3 차원은 없음
scores.permute(1, 2, 2)         # RuntimeError예외 발생: 차원 중복(누락도 발생)
scores.transpose(1, 3)          # IndexError 예외 발생
scores.transpose(1, 1)          # 예외 발생하지 않음(차원 순서도 그대로)
```
