## 연습 문제 1-1

> 파이토치 프레임워크를 통해 딥러닝 모델 개발에 사용할 수 있는 여러 데이터셋을 쉽게 구할 수 있다.
> 다음 코드는 MNIST 데이터셋을 불러와 `images`와 `labels` 두 텐서를 생성한다. (코드 1-22)
> 이렇게 생성한 `images`와 `labels` 두 텐서의 형태를 출력해 보자.

```python
# 코드 1-22 MNIST 데이터셋 불러오기

from torchvision import datasets, transforms

# 원문 코드의 root='./data' 대신 저장소의 download 디렉터리를 사용
transform = transforms.ToTensor()
train_set = datasets.MNIST(root='../../download', train=True, download=True,
                           transform=transform)
train_loader = torch.utils.data.DataLoader(dataset=train_set, batch_size=32,
                                           shuffle=True)
images, labels = next(iter(train_loader))
```

여기서 나오는 `DataLoader`, `transform` 등은 4장에서 다룬다. 지금은 실행 결과 텐서의 형태만 확인하면 된다.

**텐서의 형태 확인**
- `shape` 속성으로 확인
- 또는 `size()` 메서드로도 확인 가능

```python
print(f'images 텐서의 형태: {images.shape}')
print(f'labels 텐서의 형태: {labels.size()}')
```

**두 텐서와 그 형태에 대하여**

`images`는 `(32, 1, 28, 28)`, `labels`는 `(32,)` 형태다.

**`images` 텐서**: 32장의 이미지 뭉치 데이터 텐서
- 첫 번째 차원: 32장의 이미지로 구성된 이미지 뭉치 데이터(배치 데이터)
- 두 번째 차원: 회색조 이미지로 하나의 색 채널만 사용
- 세 번째, 네 번째 차원: 각 이미지는 28x28의 크기로, 각 픽셀값이 저장

**'labels` 텐서**: 32장 이미지의 정답 뭉치 데이터 텐서
- 하나의 차원으로 구성되며, 각 차원에는 0에서 9까지 정답이 저장

```python
# 두 텐서에 저장된 첫 번째 이미지의 이미지 텐서와 정답 텐서를 슬라이싱/인덱싱한 후 출력
print(f'첫 번째 이미지 텐서의 형태: {images[0].shape}')
print(f'첫 번째 이미지의 정답: {labels[0]}')
```

참고: **텐서의 차원 확인**
- ndim 속성 또는 dim() 메서드로 확인

```python
print(f'images 텐서의 차원: {images.ndim}')
print(f'labels 텐서의 차원: {labels.dim()}')
```

## 연습 문제 1-2

> [연습 문제 1-1]에서 생성한 텐서 중 `images` 텐서에는 가로, 세로 각각 28픽셀로 구성된 32장의 이미지 정보가
> 저장되어 있다. 이 텐서의 두 번째 차원은 컬러 이미지의 색상별 데이터로 구성된 색상 채널 차원인데,
> MNIST 데이터와 같이 색상 채널이 하나뿐인 회색조 이미지에서는 색상 채널 차원이 불필요하다.
> `images` 텐서에서 색상 채널 차원을 제거해 보자.

**텐서의 차원 삭제**
- `squeeze()` 메서드로 제거
- 크기가 1인 차원만 제거

```python
# 색상 채널은 두 번째 차원이며, 두 번째 차원의 인덱스는 1
squeezed = images.squeeze(dim=1)
print(f'제거 전: {images.shape}')
print(f'제거 후: {squeezed.shape}')

# 인자를 생략하면 크기 1인 모든 차원 제거(결과는 동일)
print(f'dim 생략: {images.squeeze().shape}')

# 크기가 1이 아닌 차원은 삭제되지 않음(예외도 발생하지 않음)
# 인자의 dim= 생략 가능
print(f'첫 번째 차원 제거 시도: {images.squeeze(0).shape}')
```

**이미지 한장 뿐인 `(1, 1, 28, 28)` 형태의 텐서의 경우**

`dim` 인자를 생략하면 결과가 달라짐

```python
last_batch = images[:1]
print(f'이미지 한 장: {last_batch.shape}')
print(f'squeeze():  {last_batch.squeeze().shape}')    # 배치 차원까지 사라진다
print(f'squeeze(1): {last_batch.squeeze(1).shape}')   # 색상 채널만 제거된다
```

`dim` 인자를 지정하지 않으면 의도와 다르게 엉뚱한 채널까지 지워버릴 수 있으므로 `dim` 인자를 지정하는 것이 안전하다!

## 연습 문제 1-3

> [연습 문제 1-2]에서 색상 채널을 제거한 텐서의 마지막 두 차원은 각각 가로 픽셀과 세로 픽셀에 해당하는
> 차원이다. 이 두 차원의 데이터를 이어 붙여 각 이미지가 크기 784의 1차원 텐서로 표현되도록
> `images` 텐서를 변형해 보자.

**텐서의 형태 변형**
- `(32, 28, 28)` 텐서를 `(32, 784)` 텐서로 변형
- 텐서의 형태를 바꿀 때는 `reshape()` 메서드 사용

```python
# reshape() 메서드를 사용
print(f'평탄화 전: {squeezed.shape}')
flattened = squeezed.reshape(32, 784)
print(f'평탄화 후: {flattened.shape}')
```

**평탄화 flatten() 메서드**
- 형태 변형 중 둘 이상의 차원을 이어 붙이는 툭수한 형태의 변형
- `flatten()` 메서드를 이어 붙일 차원 인자와 함께 사용

```python
# flatten() 메서드를 사용
# 첫 번째 차원(이미지 번호)은 그대로 두고, 두 번째 차원부터 끝까지 이어 붙인다
flattened = squeezed.flatten(1, 2)
print(f'평탄화 전: {squeezed.shape}')
print(f'평탄화 후: {flattened.shape}')
```

```python
# 음수 인덱스로 지정 가능: 두 번째 차원부터 마지막 차원까지
print(f'평탄화 후: {squeezed.flatten(1, -1).shape}')

# 인자를 지정하지 않으면?
print(f'1차원으로 평탄화: {squeezed.flatten().shape}')
```

**색상 채널 제거까지 한 번에 처리**

```python
# images: 원래의 이미지 뭉치 텐서, (32, 1, 28, 28) 형태
print(f'한 번에 처리: {images.flatten(1, 3).shape}')
```

**배치 크기가 바뀌었을 때의 reshape() 메서드**

- 배치 크기가 16으로 바뀌었을 때 `reshape(32, 784)`는 예외가 발생!
- `reshape()` 메서드의 인자로 지정하는 텐서 형태는 원래의 텐서와 요소 수가 일치해야 함

```python
small = squeezed[:16]     # 배치 크기가 16으로 바뀐 경우
try:
    small.reshape(32, 784)
except RuntimeError as e:
    print(f'reshape(32, 784) 오류: {e}')
print(f'flatten(1, -1):       {small.flatten(1, -1).shape}')
```

- `flatten()` 메서드를 사용하는 쪽이 안전
- 평탄화 변환이 아니거나, 굳이 `reshape()` 메서드를 사용해야 한다면?

```python
batch_size = squeezed.shape[0]          # 첫 번째 차원(배치)의 크기
flattened = squeezed.reshape(batch_size, 784)
print(f'평탄화 후: {flattened.shape}')
```
